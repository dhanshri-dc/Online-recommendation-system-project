# MODEL BUILDING PART 

##  Feature Engineering

### Binary Encoding

```python
df['certification_offered'] = df['certification_offered'].map({
    'Yes':1,
    'No':0
})

df['study_material_available'] = df['study_material_available'].map({
    'Yes':1,
    'No':0
})
```

```python
df
```

### Difficulty Level Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['difficulty_level'] = le.fit_transform(df['difficulty_level'])
```

```python
df
```

### Feature Selection

```python
sample_df = df.sample(
    5000,
    random_state=42
)
sample_df = sample_df.reset_index(drop=True)
```

```python
sample_df.head()
```

### FEATURE SELECTION

```python
features = sample_df[[
    'course_duration_hours',
    'certification_offered',
    'difficulty_level',
    'rating',
    'enrollment_numbers',
    'course_price',
    'feedback_score',
    'study_material_available',
    'time_spent_hours',
    'previous_courses_taken'
]]
```

```python
features.head()
```

### FEATURE SCALING

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)
```

```python
print(scaled_features)
```

```python
## CONVERT ARRAY TO DATAFRAME

scaled_df = pd.DataFrame(
    scaled_features,
    columns=features.columns
)

print(scaled_df.head())
```

### COSINE SIMILARITY

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(
    scaled_features
)
```

```python
similarity[:5]
```

```python
similarity_df = pd.DataFrame(similarity)
```

```python
similarity_df.head()
```

### RECOMMENDATION FUNCTION

```python
def recommend_courses(course_name):

    # COURSE INDEX

    course_index = sample_df[
        sample_df['course_name'] == course_name
    ].index[0]

    # SIMILARITY SCORES

    similarity_scores = list(
        enumerate(similarity[course_index])
    )

    # SORTING

    sorted_courses = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # RECOMMENDATIONS

    recommended_courses = []

    for i in sorted_courses[1:6]:

        recommended_courses.append(
            sample_df.iloc[i[0]]['course_name']
        )

    return recommended_courses
```

### TEST THE MODEL

```python
print(
    recommend_courses(
        sample_df['course_name'].iloc[0]
    )
)
```

```python
sample_df['course_name'].head(10)
```

```python
recommend_courses("Python for Beginners")
```
